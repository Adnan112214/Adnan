```python
import os
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier, GradientBoostingClassifier
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix
from sklearn.preprocessing import StandardScaler
import glob
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import LSTM, Dense, Dropout
from tensorflow.keras.utils import to_categorical
from sklearn.preprocessing import LabelEncoder
import matplotlib.pyplot as plt
from sklearn.preprocessing import MinMaxScaler
import seaborn as sns
import os
from tensorflow.keras.callbacks import EarlyStopping
import matplotlib.pyplot as plt
from statsmodels.tsa.arima.model import ARIMA
from statsmodels.tsa.stattools import adfuller
from sklearn.metrics import mean_squared_error
from sklearn.metrics import mean_absolute_error
from prophet import Prophet
```

```python

def read_multiple_txt_files(root_folder_path):

    all_data = []

    columns = [
        "datetime",
        "A_Leq", "A_L10", "A_L95",
        "C_Leq", "C_L10", "C_L95"
    ]

    for dirpath, dirnames, filenames in os.walk(root_folder_path):

        for file in filenames:

            if file.endswith('.txt'):

                full_file_path = os.path.join(dirpath, file)

                try:
                    # Read txt file, ignore comments
                    df = pd.read_csv(
                        full_file_path,
                        comment="#",
                        header=None,
                        names=columns,
                        sep=",",
                        skipinitialspace=True
                    )

                    # Convert datetime
                    df["datetime"] = pd.to_datetime(
                        df["datetime"],
                        format="%d/%m/%Y %H:%M:%S",
                        errors="coerce"
                    )

                    
                    df["source_file"] = file

                    # Drop invalid rows
                    df = df.dropna()

                    all_data.append(df)

                except Exception as e:
                    print(f"Error reading {full_file_path}: {e}")

    # Combine all files into one DataFrame
    if len(all_data) > 0:
        final_df = pd.concat(all_data, ignore_index=True)
    else:
        final_df = pd.DataFrame()

    return final_df


root_folder = "2013"

noise_df = read_multiple_txt_files(root_folder)

# Show result
print(noise_df.head())
print(noise_df.info())

# Save for ML
noise_df.to_csv("processed_noise_data.csv", index=False)
```

# Classification Problem

```python

#Function definition 
# # Convert A_Leq into binary classification (Peak vs Normal)
# df (DataFrame): Input noise dataframe
#percentile (float): Threshold percentile (default = 0.90 → top 10% peaks)
#df (DataFrame): Updated dataframe with noise_class column
#threshold (float): Calculated threshold value


import pandas as pd

def create_noise_class(df, percentile=0.90, balance=True):
    
    
    # Calculate threshold
    threshold = df["A_Leq"].quantile(percentile)
    
    #Create binary class
    df["noise_class"] = (df["A_Leq"] > threshold).astype(int)
    
    # Step 3: Balance classes 
    if balance:
        class_0 = df[df["noise_class"] == 0]
        class_1 = df[df["noise_class"] == 1]
        
        
        class_0_downsampled = class_0.sample(n=len(class_1), random_state=42)
        
        # Combine
        df = pd.concat([class_0_downsampled, class_1])
        
        # Shuffle
        df = df.sample(frac=1, random_state=42).reset_index(drop=True)
    
    return df, threshold



noise_df, threshold = create_noise_class(noise_df, percentile=0.90, balance=True)


print(f"Threshold (90th percentile): {threshold:.2f}")

print("\nClass Distribution:")
print(noise_df["noise_class"].value_counts())

print("\nClass Ratio:")
print(noise_df["noise_class"].value_counts(normalize=True))

noise_df.to_csv("classified_data.csv", index=False)
```

# ARIMA

```python
# Sort by time (VERY IMPORTANT)
noise_df = noise_df.sort_values("datetime")

# Train-test split (time-based)
train_size = int(len(noise_df) * 0.8)

train = noise_df.iloc[:train_size]
test = noise_df.iloc[train_size:]

# Fit ARIMA (you can tune order later)
model = ARIMA(train["A_Leq"], order=(2,1,2))
model_fit = model.fit()

# Forecast
forecast = model_fit.forecast(steps=len(test))

# Use same threshold from training
threshold = train["A_Leq"].quantile(0.90)

# True labels
y_true = (test["A_Leq"] > threshold).astype(int)

# Predicted labels (from ARIMA forecast)
y_pred = (forecast > threshold).astype(int)
```

## Evaluation of ARIMA

```python

tn, fp, fn, tp = confusion_matrix(y_true, y_pred).ravel()

print(f"TP: {tp}, TN: {tn}, FP: {fp}, FN: {fn}")

# Accuracy
accuracy = (tp + tn) / (tp + tn + fp + fn)

# Miss Rate
miss_rate = fn / (fn + tp) if (fn + tp) > 0 else 0

# Specificity
specificity = tn / (tn + fp) if (tn + fp) > 0 else 0

# Precision
precision = tp / (tp + fp) if (tp + fp) > 0 else 0

# Recall
recall = tp / (tp + fn) if (tp + fn) > 0 else 0

print("\n===== Performance Metrics =====")
print(f"Accuracy     : {accuracy:.4f}")
print(f"Miss Rate    : {miss_rate:.4f}")
print(f"Specificity  : {specificity:.4f}")
print(f"Precision    : {precision:.4f}")
print(f"Recall       : {recall:.4f}")

cm = confusion_matrix(y_true, y_pred)

# Extract values
tn, fp, fn, tp = cm.ravel()

# Print raw matrix
print("Confusion Matrix (raw):")
print(cm)

# Pretty format
cm_df = pd.DataFrame(
    cm,
    index=["Actual Normal (0)", "Actual Peak (1)"],
    columns=["Predicted Normal (0)", "Predicted Peak (1)"]
)

plt.figure()

# Plot heatmap
plt.imshow(cm)

# Labels
classes = ["Normal (0)", "Peak (1)"]
plt.xticks(np.arange(len(classes)), classes)
plt.yticks(np.arange(len(classes)), classes)

# Title and axis labels
plt.title("Confusion Matrix Heatmap (ARIMA Classification)")
plt.xlabel("Predicted Label")
plt.ylabel("True Label")

# Add color bar
plt.colorbar()

# Annotate each cell
for i in range(cm.shape[0]):
    for j in range(cm.shape[1]):
        plt.text(j, i, cm[i, j],
                 ha="center", va="center")

plt.tight_layout()
plt.show()
```

# Prophet

```python
noise_df = noise_df.sort_values("datetime")

# Train-test split
train_size = int(len(noise_df) * 0.8)
train = noise_df.iloc[:train_size]
test = noise_df.iloc[train_size:]

# Rename for Prophet
prophet_train = train[["datetime", "A_Leq"]].rename(
    columns={"datetime": "ds", "A_Leq": "y"})

model = Prophet(daily_seasonality=True)
model.fit(prophet_train)

# Create future dataframe
future = model.make_future_dataframe(periods=len(test), freq="5min")

forecast = model.predict(future)

# Extract only test predictions
yhat = forecast["yhat"].iloc[-len(test):].values

# Use SAME threshold from training
threshold = train["A_Leq"].quantile(0.90)

# True labels
y_true = (test["A_Leq"] > threshold).astype(int)

# Predicted labels
y_pred = (yhat > threshold).astype(int)
```

## Evaluation of Prophet

```python
cm = confusion_matrix(y_true, y_pred)

tn, fp, fn, tp = cm.ravel()

print("Confusion Matrix (raw):")
print(cm)

# Heatmap
plt.figure()
plt.imshow(cm)

classes = ["Normal (0)", "Peak (1)"]
plt.xticks(np.arange(2), classes)
plt.yticks(np.arange(2), classes)

plt.title("Confusion Matrix Heatmap (Prophet Classification)")
plt.xlabel("Predicted")
plt.ylabel("Actual")

plt.colorbar()

# Annotate
for i in range(2):
    for j in range(2):
        plt.text(j, i, cm[i, j], ha="center", va="center")

plt.tight_layout()
plt.show()


# Accuracy
accuracy = (tp + tn) / (tp + tn + fp + fn)

# Miss Rate
miss_rate = fn / (fn + tp) if (fn + tp) > 0 else 0

# Specificity
specificity = tn / (tn + fp) if (tn + fp) > 0 else 0

# Precision
precision = tp / (tp + fp) if (tp + fp) > 0 else 0

# Recall
recall = tp / (tp + fn) if (tp + fn) > 0 else 0

print("\n===== Prophet Performance =====")
print(f"Accuracy     : {accuracy:.4f}")
print(f"Miss Rate    : {miss_rate:.4f}")
print(f"Specificity  : {specificity:.4f}")
print(f"Precision    : {precision:.4f}")
print(f"Recall       : {recall:.4f}")
```

# LSTM

```python
df = pd.read_csv("classified_data.csv")

# Convert datetime and sort
df["datetime"] = pd.to_datetime(df["datetime"])
df = df.sort_values("datetime")

features = ["A_Leq", "A_L10", "A_L95", "C_Leq", "C_L10", "C_L95"]
target = "noise_class"

X = df[features].values
y = df[target].values

scaler = MinMaxScaler()
X_scaled = scaler.fit_transform(X)

def create_sequences(X, y, window_size=10):
    X_seq, y_seq = [], []
    
    for i in range(len(X) - window_size):
        X_seq.append(X[i:i+window_size])
        y_seq.append(y[i+window_size])
    
    return np.array(X_seq), np.array(y_seq)

window_size = 10
X_seq, y_seq = create_sequences(X_scaled, y, window_size)

split = int(0.8 * len(X_seq))

X_train, X_test = X_seq[:split], X_seq[split:]
y_train, y_test = y_seq[:split], y_seq[split:]

model = Sequential([
    LSTM(64, return_sequences=True, input_shape=(window_size, len(features))),
    Dropout(0.2),
    
    LSTM(32),
    Dropout(0.2),
    
    Dense(1, activation='sigmoid')
])

model.compile(
    optimizer='adam',
    loss='binary_crossentropy',
    metrics=['accuracy']
)

model.summary()

model.fit(
    X_train, y_train,
    epochs=10,
    batch_size=32,
    validation_split=0.1,
    verbose=1
)
```

```python
y_pred_prob = model.predict(X_test)
y_pred = (y_pred_prob > 0.5).astype(int).flatten()

cm = confusion_matrix(y_test, y_pred)
tn, fp, fn, tp = cm.ravel()

print("Confusion Matrix:")
print(cm)

# Heatmap
plt.figure()
plt.imshow(cm)

classes = ["Normal (0)", "Peak (1)"]
plt.xticks(np.arange(2), classes)
plt.yticks(np.arange(2), classes)

plt.title("Confusion Matrix Heatmap (LSTM)")
plt.xlabel("Predicted")
plt.ylabel("Actual")

plt.colorbar()

for i in range(2):
    for j in range(2):
        plt.text(j, i, cm[i, j], ha="center", va="center")

plt.tight_layout()
plt.show()
# Accuracy
accuracy = (tp + tn) / (tp + tn + fp + fn)

# Miss Rate
miss_rate = fn / (fn + tp) if (fn + tp) > 0 else 0

# Specificity
specificity = tn / (tn + fp) if (tn + fp) > 0 else 0

# Precision
precision = tp / (tp + fp) if (tp + fp) > 0 else 0

# Recall
recall = tp / (tp + fn) if (tp + fn) > 0 else 0

print("\n===== LSTM Performance =====")
print(f"Accuracy     : {accuracy:.4f}")
print(f"Miss Rate    : {miss_rate:.4f}")
print(f"Specificity  : {specificity:.4f}")
print(f"Precision    : {precision:.4f}")
print(f"Recall       : {recall:.4f}")
```
